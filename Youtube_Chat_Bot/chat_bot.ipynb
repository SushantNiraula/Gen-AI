## 1. Step 1 ***Extract the transcript***

```python
this_video_id='Gfr50f6ZBvo'
try:
    transcript_list=YouTubeTranscriptApi.get_transcript(this_video_id,languages=['en'])

    transcript= " ".join(chunk['text'] for chunk in transcript_list)
    print(transcript)

except TranscriptsDisabled:
    print("No captions available for this video.")
```

## 2. Step 2 ***Indexing or Text Splitting***

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter= RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks=splitter.create_documents([transcript])
```

```python
len(chunks)
```

```python
chunks[0].page_content
```

## 3. Step 3 ***Indexing(Embedding Generation and storing in vector store)***

```python
 AZURE_OPENAI_ENDPOINT='https://learning-sushant.openai.azure.com/'
 AZURE_OPENAI_API_VERSION='2024-02-01'
 AZURE_OPENAI_DEPLOYMENT_NAME='text-embedding-3-large'
 AZURE_OPENAI_API_KEY=os.environ['AZURE_EMBEDDING_API_KEY']
```

```python
from langchain_openai import AzureOpenAIEmbeddings
from langchain_community.vectorstores import FAISS
embeddings= AzureOpenAIEmbeddings(azure_endpoint=AZURE_OPENAI_ENDPOINT,azure_deployment=AZURE_OPENAI_DEPLOYMENT_NAME,openai_api_version=AZURE_OPENAI_API_VERSION,api_key=AZURE_OPENAI_API_KEY)
vector_store= FAISS.from_documents(chunks, embeddings)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(['9f89ad96-a3a4-49e3-80b2-150adc0cde89'])
```

## Building a Retriever

```python
retriever=vector_store.as_retriever(search_type='similarity',search_kwargs={'k':4})
```

```python
retriever
```

```python
retriever.invoke('What is deepmind')
```

## Augmentation

```python
from langchain_core.prompts import PromptTemplate
prompt= PromptTemplate(
    template='''
    you are a helpful assistant.
    Answer only from the provided transcript context.
    If the context is insufficient, just say you don't know.
    {context}
    Question: {question}''',
    input_variables=['context','question']

)
```

```python
question='is the topic of aliens discussed in the video? If yes then what is discussed'
retrieved_docs = retriever.invoke(question)
```

```python
retrieved_docs
```

```python
context_text= "\n\n".join(doc.page_content for doc in retrieved_docs)
```

```python
final_prompt= prompt.invoke({'context':context_text, 'question':question})
```

## Generation.

```python
answer=chat_model.invoke(final_prompt)
print(answer.content)
```

## Building Chains!!!


```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_docs(retrieved_docs):
    context_text= "\n\n".join(doc.page_content for doc in retrieved_docs)
    return context_text
```

```python
parallel_chain=RunnableParallel({
    'context': retriever | RunnableLambda(format_docs),
    'question': RunnablePassthrough(),
}
)
```

```python
parallel_chain.invoke('who is Demis')
```

```python
parser= StrOutputParser()
```

```python
main_chain= parallel_chain | prompt | chat_model | parser

```

```python
main_chain.invoke('Can you summerize the video')
```
