## Custom Tools
Use them when:
* You want to call ***your own API's***
* You want to ***encapsulate business logic***
* You want the LLM to interact with ***your database, product, or app***

```python
from langchain_core.tools import tool
from torch.optim.optimizer import required
```

```python
def multiply(a,b):
    """Multiply two numbers."""
    return a*b
```

```python
## step 2 - add the type hints:
def multiply(a: int, b: int)-> int:
    """Multiply two numbers."""
    return a*b
```

```python
## step 3- add tool decorator
@tool
def multiply(a: int, b: int)-> int:
    """Multiply two numbers."""
    return a*b
```

```python
result= multiply.invoke({"a":2, 'b':5})
print(result)
```

```python
type(multiply)
```

```python
print(multiply.description)
print(multiply.args)
print(multiply.name)
```

Ways to create Tools:
1. using @tool decorator
2. using StructuredTool and Pydantic
3. Using BaseTool class

A Structured Tool in Langchain is a special type of toolw here the input to the tool follows a structured schema, typically defined using a Pydantic Model.

```python
from langchain.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyTool(BaseModel):
    a: int = Field(required=True,description='The first number to multiply')
    b: int= Field(required=True, description='The second number to mutliply')
```

```python
def multiply_func(a:int,b:int)->int:
    return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name='multiply',
    description='Multiply two numbers',
    args_schema=MultiplyTool
)
```

```python
result=multiply_tool.invoke({"a":2,"b":5})
```

```python
print(result)
print(multiply_tool.name)
print(multiply_tool.description)
```

## BaseTool
BaseTool is the abstract base class for all tools in LangChain. It defines the core structure and interface that any tool must follow, whether it's a simple one-liner or fully customized function.

All other tool types like @

```python
## Using BaseTool class
from langchain.tools import BaseTool
from typing import Type
```

```python
class MultiplyInput(BaseModel):
    a: int= Field(required=True, description='The first number to add')
    b: int= Field(required=True, description='The second number to add')

```

```python
class MultiplyTool(BaseTool):
    name:str='multiply'
    description:str='Multiply two numbers'
    args_schema:Type[BaseModel]=MultiplyInput
    def _run(self, a:int,b:int)->int:
        return a*b
```

```python
multiply_tool=MultiplyTool()
```

```python
result= multiply_tool.invoke({'a':10, 'b':3})
```

```python
print(result)
print(multiply_tool.name)
print(multiply_tool.description)
```

## Toolkits

A toolkit is just a collection(bundle) of related tools that serve a common purpose packaged together for convenience and reusability.

In LangChain:
* A toolkit might be : GoogleDriveToolkit
* And it can contain the following tools

* `GoogleDriveCreateFileTool:` Uploada file
* `GoogleDriveSearchTool:` Search for a file by name/content
* `GoogleDriveReadFileTool:` Read contents of a file

```python
@tool
def add(a:int, b:int)-> int:
    """Add two numbers."""
    return a+b
@tool
def multiply(a:int, b:int)-> int:
    """Multiply two numbers."""
    return a*b
```

```python
class MathToolkit:
    def get_tools(self):
        return[add, multiply]
```

```python
toolkit=MathToolkit()
tools=toolkit.get_tools()
```

```python
for tool in tools:
    print(tool.name)
```
